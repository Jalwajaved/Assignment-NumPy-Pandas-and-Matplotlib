# Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("✅ Libraries Imported Successfully.")
```

# NumPy Task 1 (Statistics)
**Task:** Create an array 1-20. Compute Mean, Max, and Standard Deviation.

```python
# 1. Create Array from 1 to 20
arr = np.arange(1, 21)
print(f"Array: {arr}")

# 2. Compute Statistics
mean_val = np.mean(arr)
max_val = np.max(arr)
std_val = np.std(arr)

print(f"\n📊 NumPy Statistics:")
print(f"   Mean: {mean_val}")
print(f"   Maximum: {max_val}")
print(f"   Standard Deviation: {std_val:.2f}")
```

# NumPy Task 2 (Boolean Masking)
**Task:** Create a 5x5 random matrix and replace values < 0.5 with 0.

```python
# 1. Create 5x5 Random Matrix (values between 0 and 1)
np.random.seed(42) # Seed ensures the result is the same every time you run it
matrix = np.random.rand(5, 5)

print("Original Random Matrix:\n")
print(matrix.round(2))

# 2. Replace values < 0.5 with 0
# Logic: We use a 'Mask' (boolean indexing)
matrix[matrix < 0.5] = 0

print("\nModified Matrix (Values < 0.5 become 0):\n")
print(matrix.round(2))
```

# Pandas Task 3 (Load & Calculate)
**Task:** Load CSV, show first 5 rows, and calculate the average of one column.

```python
# 1. Load the Dataset
df = pd.read_csv('/content/Train.csv')

# 2. Display First 5 Rows
print("First 5 Rows of Data:")
display(df.head())

# 3. Calculate Average of 'Age'
# We use the 'Age' column as it is definitely numerical.
average_age = df['Age'].mean()

print(f"\n📈 Average Age of Customers: {average_age:.2f} years")
```

# Pandas Task 4 (Filter & Save)
**Task:** Filter rows where a column exceeds a threshold and save to a new CSV.

```python
# 1. Define Threshold
# Let's filter for people older than 50
threshold = 50
filtered_df = df[df['Age'] > threshold]

print(f"Filter Criteria: Age > {threshold}")
print(f"Rows before filter: {len(df)}")
print(f"Rows after filter:  {len(filtered_df)}")

# 2. Save to New CSV
output_filename = "Senior_Customers.csv"
filtered_df.to_csv(output_filename, index=False)

print(f"\n✅ Saved filtered data to '{output_filename}'")
```

# Matplotlib Task 5 (Line Graph)
**Task:** Plot a line graph of values 1-10 with labels.

```python
# Data
x = range(1, 11)
y = [i**2 for i in x] # Let's plot squares (1, 4, 9...) so the line curves nicely

plt.figure(figsize=(8, 4))
plt.plot(x, y, marker='o', linestyle='-', color='blue', label='y = x^2')

# Labels
plt.xlabel("Input Values (1-10)")
plt.ylabel("Squared Values")
plt.title("Line Graph of Squares")
plt.legend()
plt.grid(True)

plt.show()
```

# Matplotlib Task 6 (Bar Chart)
**Task:** Create a bar chart for 5 categories

```python
# Data: Let's use 5 professions from your dataset
categories = ['Engineer', 'Doctor', 'Artist', 'Lawyer', 'Homemaker']
values = [45, 30, 60, 25, 40] # Hypothetical count

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color=['#FF5733', '#33FF57', '#3357FF', '#FF33A1', '#33FFF5'])

# Labels
plt.xlabel("Profession")
plt.ylabel("Count")
plt.title("Number of People per Profession")

plt.show()
```
